```python
# useful for visualisation
import nglview as nv
import numpy as np

def aviz(crystal_structure, atom_radius=0.3, bonds=True):

    view = nv.show_pymatgen(crystal_structure)

    view.clear_representations()

    # --- simple, reliable logic ---
    if bonds:
        view.add_ball_and_stick(radius=atom_radius)
    else:
        view.add_spacefill(radius=atom_radius)

    view.add_unitcell()

    

    view.camera = "orthographic"
    view.center()


    return view
```

# Building computational crystal materials models


Often in computational materials science, certainly when we want to do an atomistic calculation, e.g. DFT, or simulation, eg. MD, we need to start from a materials model that described the unit cell and the positions and  types of the atoms. As we learned in class if you took NSCI0032 and other modules, solid state materials are often described by certain symmetry space groups. See for example [chapter 1 of West](https://ucl.primo.exlibrisgroup.com/permalink/44UCL_INST/2mbi31/alma9933003044604761), which is available in the UCL library (online).  This is an excellent resource that also has a review of the basic materials structures that you will be working, e.g. Rocksalt structures, or perovskites etc. So it is a good read. You may also consult the book of Atkins. 

One of the most powerful ways to construct such atomistic models, is to follow the, what can be called, the "Wyckoff method" where by we need to know the space group, which will determine the symmetry and lattice parameters, then we need to look for the so called Wyckoff Positions, which are high symmetry special positions within the group. These are given usually a multiplicity and generating sites in addition to symmetry. This may sound complex now, but when you follow the tutorial you will see that it is very simple and practical means to get models for which you have full understanding and control. 

As alternative, you may see sometimes papers in which they simply download ready models from some database, e.g. materials project or the COD (Crystallography Openn DataBase, or ICSD, see below), but these will not provide as much fundamental understanding and insight to the structure and how it was build. This however, is important for materials discovery! so we go the Wyckoff way here! and you are strongly encouraged to do so. 

The steps are as follows: 

1. First read and familiarise yourself with e.g. chapter 1 of West (as above), it also has a very good description of the common materials crystal systems, like Perovskites, rock salt etc, which many students are working on any way. Pay attention to the description of the Wyckoff structures.

## Concrete example: Building a NASICON type crystal Structure ($LiTi_2(PO_4)_3$)

This tutorial demonstrates how to build a NASICON-type crystal structure, $LiTi_2(PO_4)_3$, using our tool and pymatgen. We will construct the crystal lattice and atomic structure from experimentally reported crystallographic data, generate a pymatgen `Structure` object using `Lattice` and `Structure.from_spacegroup()`, and finally create a Quantum ESPRESSO input file using `PWInput`.

```python
# First some imports
import numpy as np
from pymatgen.core import Lattice, Structure
from pymatgen.io.pwscf import PWInput
import os

# It is good to know where are we working!
os.getcwd()
os.chdir("/home/jovyan/crystpack/NSCI0032_25-26/Pymatgen_building_lattice") # you need to replace it with your own path
os.getcwd()
```

## Define Crystal Structure

The crystal structure data for our LiTi₂(PO₄)₃ system is shown below as obtaioned from teh paper and ICSD as explained in the introduction above. This one is namely from [(Aatiq et al., 2002)](https://doi.org/10.1039/B203652P). Below we put the original table from the publication, which should be the same as the table above from the ICSD. 


<center>
  <img src="figures/image8.jpg" width="1000">

  <div>
    Figure 1. Crystal structure of LiTi₂(PO₄)₃.
  </div>
</center>

<br>

This structure belongs to the $ R\bar{3}c $ space group, corresponding to space group number 167. We provide the lattice parameters together with the elemental information and fractional coordinates of the Wyckoff positions, and then use `Structure.from_spacegroup()` in pymatgen to automatically generate the full crystal structure.


```python
a_hex = 8.511
c_hex = 20.843
```

Note the system is hexagonal, as can be seen both from ICSD, Wyckoff table (the bilbao one), and the Figure 1 from the publication.

Check the documentaion of pymatgen for more options and other crystal systems: https://pymatgen.org/pymatgen.core.html#module-pymatgen.core.lattice 

```python
ltp_lattice = Lattice.hexagonal(a_hex, c_hex)
spacegroup = 167 # this is from the tables above

# these are the wyckoff sites: Li (6b), then Ti (12 c), P (18 e), O (36 f), O (36 f)
# note these are not the actual positions of all atoms, but the gnerating sites.
species = ['Li', 'Ti', 'P', 'O','O']

coords = [
    [0, 0, 0],      # the Li 6 b generating site from teh table etc
    [0, 0, 0.1421],  # Ti (12c)
    [0.2903,0,0.25], # P (18 e)
    [0.1850,0.9960,0.1901], # O1 (36 f )
    [0.1884,0.1641,0.0810] # O2 (36 f )
]

# now we tell pymatgen to create the structure for us:
ltp = Structure.from_spacegroup(spacegroup,ltp_lattice, species,coords)
```

```python
aviz(ltp)
```

# Quantum Espresso Input
Next, we define the key parameters required for the Quantum ESPRESSO input file, including pseudopotentials, calculation type (e.g. relax), plane-wave energy cutoffs, K-point sampling, convergence criteria, and other simulation settings.

```python
# Pseudopotentials
# all Pseudo potentials are already available to you under /public/adham/qe/pp/upf/, you need to copy them locally. 
pseudos = {
    "Li": "li_pbesol_v1.4.uspp.F.UPF",
    "Ti": "ti_pbesol_v1.4.uspp.F.UPF",
    "P": "P.pbesol-n-rrkjus_psl.1.0.0.UPF",
    "O": "O.pbesol-n-kjpaw_psl.0.1.UPF",
}
```

```python
# QE parameters
control = {
    "calculation": "relax",
    "prefix": "ltp",
    "pseudo_dir": "/public/adham/qe/pp/upf/",
    "outdir": "./out",
}

system = {
    "ecutwfc": 35,
    "ecutrho": 400,
    "occupations": "smearing",
    "smearing": "gaussian",
    "degauss": 0.02,
}

electrons = {
    "conv_thr": 1e-6,
    "mixing_beta": 0.3,
}
```

```python
# Build PWInput.. we use the pymatgen class PWInput: https://pymatgen.org/pymatgen.io.html#module-pymatgen.io.pwscf
pw = PWInput(
    structure=ltp,
    pseudo=pseudos,
    control=control,
    system=system,
    electrons=electrons,
    kpoints_grid=(3, 3, 1),
)


# Write file
pw.write_file("ltp.in")

# now open the file and check the input parameters. 
```
