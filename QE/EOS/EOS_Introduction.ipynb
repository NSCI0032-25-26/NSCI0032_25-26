# The Equation of State (EOS)

- The EOS allows us to  to find the equilibrium lattice parameters

- We now know how to define a supercell for a DFT calculation for a material, e.g.,  with the simple cubic crystal structure (e.g. see the  tutorial from last week). 

- In this tutorial we assume that we have a DFT code that can give us the total energy of  the system. 

- This tutorial adresses teh question: *How can we use these calculations to determine the lattice constants of a lattice that would be observed in nature?* 



- A straight forward approach would be look how the total energy calculated from DFT for our assumed structure (say fcc or bcc etc)  as a function of the lattice constants (e.g. $a$ for fcc or bcc). Let us call this $$E_{tot}(a)$$

- Near the equilibrium lattice constant we expect this function to look like a parabola with a minimum at $a_0$.




- For $\delta a = a - a_0 \to 0$, i.e. if $\delta a$ is very small, we can expand the energy around the equilibrium lattice parameter $a_0$:

$$
E(a) \approx E_0
+ \frac12 \left.\frac{d^2E}{da^2}\right|_{a_0} (\delta a)^2,
$$

where 

$$
\left.\frac{dE}{da}\right|_{a_0} = 0
\qquad \text{and} \qquad
\left.\frac{d^2E}{da^2}\right|_{a_0} > 0
$$


- The bulk modulus is defined as

$$
B_0 = V_0 \left.\frac{d^2E}{dV^2}\right|_{V_0},
$$

- for a cubic crystal with $V = a^3$ this becomes

$$
B_0 = \frac{1}{9a_0}
\left.\frac{d^2E}{da^2}\right|_{a_0}.
$$

- Ths is only valid for a small range of lattice constants around the equilibrium value.
- B0 is response to isotropic compression, the curveture at the minimum
- Large B0 means more stiff materials
- IT is also related to PRessure (hence the name EOS!)
- More detailed mathematical treatments can give equations of state that relate these two quantities over a wider range of lattice constants. One wellknown example is the Birch–Murnaghan equation of state for isotropic solids.
- All common equations of state (Birch–Murnaghan, Vinet, Murnaghan, Rose) reduce to this quadratic form in the vicinity of the minimum.

- $B_0 = -\,V_0 \left( \frac{dP}{dV} \right)_{V_0}$
- $P(V) = -\,\frac{dE(V)}{dV}$




# The Birch-Murnaghan Equation of State

$$
E_{\text{tot}}(a)
= E_0
+ \frac{9 V_0 B_0}{16}
\left\{
\left[
\left( \frac{a_0}{a} \right)^2 - 1
\right]^3 B_0'
+
\left[
\left( \frac{a_0}{a} \right)^2 - 1
\right]^2
\left[
6 - 4 \left( \frac{a_0}{a} \right)^2
\right]
\right\}.
$$


Here:
- $a_0$ is the equilibrium lattice constant,
- $V_0$ is the equilibrium volume per atom,
- $B_0$ is the bulk modulus at zero pressure,
- $B_0' = \left( \partial B / \partial P \right)_T$.


 P Vinet et al 1989 J. Phys.: Condens. Matter 1 1941

all metals share a scaled binding-energy curve of the same functional form $E^* $
the universal equation of state (UEOS) developed by Rose, Ferrante, and Smith


- https://docs.materialsproject.org/methodology/materials-methodology/equations-of-state
- https://ase-lib.org/tutorials/eos/eos.html

$$E^*=-(1 + a^{*}) \cdot exp(-a^{*})$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Rose universal EOS (scaled)
def rose(a_star):
    return -(1 + a_star) * np.exp(-a_star)

# Domains
a_star = np.linspace(-1, 3, 400)

E_rose = rose(a_star)

plt.figure(figsize=(8,6))
plt.plot(a_star, E_rose, label="Rose Universal EOS (scaled)")
#plt.ylim(-2, 10)  # zoom in

plt.xlabel("Scaled separation (a* or r-1)")
plt.ylabel("Energy (scaled)")
plt.title("Rose EOS")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
