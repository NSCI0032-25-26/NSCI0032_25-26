```python
from ase.build import bulk
from ase.visualize import view

# Build conventional cubic diamond-Si cell (8 atoms)
a0 = 5.43  # or whatever you are using
si_conv = bulk("Si", "diamond", a=a0, cubic=True)

# Open NGL viewer
v = view(si_conv, viewer="ngl")   # v is an NGLDisplay.gui wrapper
ngl = v.view                      # this is the underlying nglview.NGLWidget

# Optional: clear default representation
ngl.clear_representations()

# --- Define the two sublattices by atom indices ---
# For ASE's conventional diamond cell, the first 4 and last 4 atoms
# correspond to the two interpenetrating FCC sublattices.
sublattice_A = [0, 1, 2, 3]
sublattice_B = [4, 5, 6, 7]

# Add ball+stick representation for each sublattice with different colors
ngl.add_representation(
    "ball+stick",
    selection=sublattice_A,   # list of atom indices
    color="red",
)
ngl.add_representation(
    "ball+stick",
    selection=sublattice_B,
    color="blue",
)
```
