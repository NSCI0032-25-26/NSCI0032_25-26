# Editing and Running LAMMPS Input Files

## 1. Opening Files
We can read, write, and edit files from from terminal, or from the menu on the left hand side. From terminal, we can run the commands: 

nano in.melt

## 2. The LAMMPS Input File

The file defines the simulation parameters for a simple Lennard-Jones melt. 

The units for the simulation are set to ``lj`` units. This is a dimensionless set of units that act as the 'natural' units for a molecular dynamics simulation.

$$
\begin{array}{|c|c|}
\hline
\textbf{Quantity} & \textbf{Reduced Form} \\
\hline
r^* & r\sigma^{-1} \\
m^* & mM^{-1} \\
t^* & t\sigma^{-1}\sqrt{\epsilon / M} \\
T^* & k_B T \epsilon^{-1} \\
E^* & E \epsilon^{-1} \\
F^* & F\sigma \epsilon^{-1} \\
P^* & P\sigma^3 \epsilon^{-1} \\
v^* & v\sqrt{M / \epsilon} \\
\rho^* & N\sigma^3 V^{-1} \\
\hline
\end{array}
$$

We can adapt various parameters to change the simulation. The choice of thermodynamic ensemble is important, as well as other parameters like starting velocities, pair coefficients for the potential, and the computation timestep.

Currently, the input file has no instruction to print the particle trajectories, which we can tell it to do by adding in the line:

dump		id all atom 5 dump.melt

Additionally, if we want thermodynamic data to be recorded more frequently, we will need to adapt the input file to reflect this.

## 2. Running LAMMPS

After making all the desired adjustments to the input file, we can now run the LAMMPS file from the terminal with the following command:

lmp -in in.melt

This will run the computation and generate a ``log.lammps`` file and a trajectory ``dump`` file.

We can take a look at the evolution of the system parameters using the ``log`` file, and plot this data. We can also use the trajectory output to view snapshots of the system at given timesteps.
